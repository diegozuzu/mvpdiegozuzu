2.1.2. Metodologia de Geração dos Dados

Processo de Aquisição:

- Amostragem: Coleta de 1.372 imagens de cédulas (50% genuínas, 50% falsas)
- Digitalização: Resolução padrão para garantir consistência nas medidas
- Processamento de Imagem:
  - Aplicação de transformada wavelet para decomposição da imagem
  - Extração de features estatísticas da transformada
  - Normalização das medidas para comparação

TRansformada Wavelet:

- Técnica: Transformada de Wavelet Contínua (CWT) ou Discreta (DWT)
- Propósito: Analisar características multiescala da textura da cédula
- Vantagens: Captura tanto informações de frequência quanto de localização espacial

Extração de Features:

Cada imagem processada gera 4 medidas estatísticas derivadas da distribuição dos coeficientes wavelet:
  - Variância: Mede a dispersão dos coeficientes
  - Assimetria (Skewness): Mede a assimetria da distribuição
  - Curtose: Mede o "achatamento" da distribuição
  - Entropia: Mede a aleatoriedade/informação

2.2. Features da Fonte de Dados

2.2.1. Esquema do Dataset

Estrutura do arquivo:

2.2.2. Descrição Técnica das Features

1. Variance (Variância)

- Nome da coluna: variance
- Tipo de dado: Float (ponto flutuante)
- Domínio esperado: [-7,0421, 6,8248] (baseado na literatura)
- Unidade de medida: Adimensional (coeficiente wavelet)
- Significado físico: Mede o grau de variação dos coeficientes wavelet. Maiores valores indicam texturas mais heterogêneas.
- Interpretação para cédulas: Cédulas genuínas tendem a ter variância mais baixa devido a processos de impressão mais controlados.
- Importância esperada: ALTA - Característica discriminativa principal

2. Skewness (Assimetria)

- Nome da coluna: skewness
- Tipo de dado: Float
- Domínio esperado: [-13,7731, 12,9516]
- Unidade: Adimensional
- Significado: Mede a assimetria da distribuição dos coeficientes. Valores positivos indicam cauda à direita, negativos cauda à esquerda.
- Interpretação: Distribuições assimétricas podem indicar imperfeições na impressão ou materiais inconsistentes.
- Importância: MÉDIA - Pode identificar padrões de falsificação específicos

3. Curtosis (Curtose)

- Nome da coluna: curtosis
- Tipo de dado: Float
- Domínio esperado: [-5,2861, 17,9274]
- Unidade: Adimensional
- Significado: Mede o "achatamento" da distribuição em relação à normal. Curtose alta = picos agudos, baixa = distribuição achatada.
- Interpretação: Cédulas genuínas podem ter curtose mais consistente devido a processos de produção padronizados.
- Importância: ALTA - Forte discriminador entre classes

4. Entropy (Entropia)

- Nome da coluna: entropy
- Tipo de dado: Float
- Domínio esperado: [-8,5482, 2,4495]
- Unidade: Bits (ou nats, dependendo da base do logaritmo)
- Significado: Mede a aleatoriedade/quantidade de informação na distribuição wavelet.
- Interpretação: Entropia baixa indica padrões regulares (genuínos), alta indica irregularidades (possíveis falsificações).
- Importância: MÉDIA-ALTA - Complementa outras medidas

5. Class (Classe/Target)

- Nome da coluna: class
- Tipo de dado: Integer (int)
- Domínio: {0, 1}
- Codificação: 0 = Cédula genuína, 1 = Cédula falsa
- Distribuição: Aproximadamente balanceada (762 vs 610)
- Significado: Rótulo supervisionado para aprendizado de máquina

2.2.3. Relações entre Features

Correlações Esperadas (Baseado no Domínio):

- Variance × Curtosis: Potencial correlação positiva moderada
- Skewness × Entropy: Possível correlação negativa fraca
- Features independentes: Cada medida captura aspectos diferentes da textura

Importância para Classificação (Hipótese):

- Variância: Principal discriminador (diferença clara entre classes)
- Curtose: Segundo melhor discriminador
- Entropia e Assimetria: Features complementares para casos ambíguos


2.2.4. Análise de Qualidade Preliminar

#3. Modelo e Catálogo de Dados

3.1 Modelagem de Dados

Para este MVP, considerando a simplicidade do dataset, utilizaremos uma modelagem otimizada para o Google Colab com abordagem Delta Lake:

3.1.1 Arquitetura de Modelagem

text
Bronze Layer (raw) → Silver Layer (cleaned) → Gold Layer (aggregated)

Tabelas Propostas:

- bronze.banknotes_raw - Dados brutos (ingestão inicial)
- silver.banknotes_cleaned - Dados tratados e enriquecidos
- gold.banknotes_metrics - Métricas agregadas por classe

3.1.2 Esquema Detalhado das Tabelas

```python
# Código para criação dos schemas no Google Colab
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *
import uuid
from datetime import datetime

# Schema para camada Bronze
bronze_schema = StructType([
    StructField("variance", FloatType(), True),
    StructField("skewness", FloatType(), True),
    StructField("curtosis", FloatType(), True),
    StructField("entropy", FloatType(), True),
    StructField("class", IntegerType(), True),
    StructField("ingestion_timestamp", TimestampType(), True),
    StructField("source_file", StringType(), True),
    StructField("batch_id", StringType(), True)
])

# Schema para camada Silver
silver_schema = StructType([
    StructField("id", StringType(), False),  # UUID único
    StructField("variance", FloatType(), True),
    StructField("skewness", FloatType(), True),
    StructField("curtosis", FloatType(), True),
    StructField("entropy", FloatType(), True),
    StructField("is_fake", IntegerType(), True),  # Renomeado de 'class' para melhor semântica
    StructField("data_quality_score", FloatType(), True),  # Score de qualidade dos dados
    StructField("is_outlier", BooleanType(), True),  # Flag para outliers
    StructField("processing_timestamp", TimestampType(), True),
    StructField("ingestion_timestamp", TimestampType(), True)
])

# Schema para camada Gold - Métricas
gold_metrics_schema = StructType([
    StructField("metric_date", DateType(), True),
    StructField("class_type", IntegerType(), True),  # 0 ou 1
    StructField("feature_name", StringType(), True),
    StructField("count", LongType(), True),
    StructField("mean", FloatType(), True),
    StructField("std_dev", FloatType(), True),
    StructField("min", FloatType(), True),
    StructField("q1", FloatType(), True),
    StructField("median", FloatType(), True),
    StructField("q3", FloatType(), True),
    StructField("max", FloatType(), True),
    StructField("outlier_count", LongType(), True)
])
```

3.2 Catálogo de Dados

3.2.1 Catálogo Completo de Tabelas

O catálogo implementa uma arquitetura Medallion de 3 camadas no Google Colab, organizada da seguinte forma:

1. CAMADA BRONZE (bronze.banknotes_raw)

- Propósito: Armazenar dados brutos "como recebidos"
- Formato: Delta Lake
- Colunas principais:
    - variance, skewness, curtosis, entropy (features originais)
    - class (0=genuína, 1=falsa)
    - Metadados: ingestion_timestamp, source_file, batch_id
- Uso: Auditoria, reprocessamento, análise forense
- Localização: /mnt/bronze/banknotes_raw

2. CAMADA SILVER (silver.banknotes_cleaned)

- Propósito: Dados limpos e validados para análise
- Transformações aplicadas:
    - Remoção de duplicatas e valores nulos
    - Renomeação: class → is_fake
    - Adição de id (UUID único)
    - Detecção de outliers (is_outlier)
    - Score de qualidade (data_quality_score)
- Colunas principais:
    - id, variance, skewness, curtosis, entropy
    - is_fake, data_quality_score, is_outlier
    - Timestamps: processing_timestamp, ingestion_timestamp
- Uso: Análise exploratória, dashboards, feature engineering
- Localização: /mnt/silver/banknotes_cleaned

3. CAMADA GOLD - PARTE A (gold.banknotes_metrics)

- Propósito: Métricas agregadas para business intelligence
- Estrutura:
    - Estatísticas por feature e classe: mean, std_dev, min, max, q1, median, q3
    - Contagem de outliers
    - Particionado por: metric_date e class_type
- Uso: Relatórios executivos, dashboards, tomada de decisão
- Localização: /mnt/gold/banknotes_metrics

4. CAMADA GOLD - PARTE B (gold.banknotes_features)

- Propósito: Features preparadas para Machine Learning
- Transformações:
    - Escalonamento (StandardScaler): variance_scaled, skewness_scaled, etc.
    - Vetorização: feature_vector (array de floats)
    - Mantém versões original e escalada
- Uso: Treinamento de modelos preditivos, inferência
- Localização: /mnt/gold/banknotes_features

```python
# Script completo de inicialização do catálogo no Google Colab
def initialize_data_catalog(spark):
    """
    Inicializa todas as tabelas no catálogo de dados
    """

    # 1. Criar banco de dados (se não existir)
    databases = ["bronze", "silver", "gold"]

    for db in databases:
        spark.sql(f"CREATE DATABASE IF NOT EXISTS {db}")
        print(f"Database {db} criado/verificado")

    # 2. Criar tabela Bronze (se não existir)
    spark.sql("""
        CREATE TABLE IF NOT EXISTS bronze.banknotes_raw (
            variance FLOAT,
            skewness FLOAT,
            curtosis FLOAT,
            entropy FLOAT,
            class INT,
            ingestion_timestamp TIMESTAMP,
            source_file STRING,
            batch_id STRING
        )
        USING DELTA
        LOCATION '/mnt/bronze/banknotes_raw'
        COMMENT 'Dados brutos de cédulas - Camada Bronze'
        TBLPROPERTIES (
            'quality' = 'bronze',
            'owner' = 'data_engineering',
            'source' = 'UCI_ML_Repository'
        )
    """)

    # 3. Criar tabela Silver com otimizações
    spark.sql("""
        CREATE TABLE IF NOT EXISTS silver.banknotes_cleaned (
            id STRING NOT NULL,
            variance FLOAT,
            skewness FLOAT,
            curtosis FLOAT,
            entropy FLOAT,
            is_fake INT,
            data_quality_score FLOAT,
            is_outlier BOOLEAN,
            processing_timestamp TIMESTAMP,
            ingestion_timestamp TIMESTAMP
        )
        USING DELTA
        LOCATION '/mnt/silver/banknotes_cleaned'
        COMMENT 'Dados tratados de cédulas - Camada Silver'
        TBLPROPERTIES (
            'quality' = 'silver',
            'owner' = 'data_engineering',
            'delta.autoOptimize.optimizeWrite' = 'true',
            'delta.autoOptimize.autoCompact' = 'true'
        )
    """)

    # 4. Criar tabela Gold para métricas
    spark.sql("""
        CREATE TABLE IF NOT EXISTS gold.banknotes_metrics (
            metric_date DATE,
            class_type INT,
            feature_name STRING,
            count BIGINT,
            mean FLOAT,
            std_dev FLOAT,
            min FLOAT,
            q1 FLOAT,
            median FLOAT,
            q3 FLOAT,
            max FLOAT,
            outlier_count BIGINT
        )
        USING DELTA
        LOCATION '/mnt/gold/banknotes_metrics'
        PARTITIONED BY (metric_date, class_type)
        COMMENT 'Métricas estatísticas por classe - Camada Gold'
        TBLPROPERTIES (
            'quality' = 'gold',
            'owner' = 'data_analytics'
        )
    """)

    # 5. Criar tabela Gold para features ML
    spark.sql("""
        CREATE TABLE IF NOT EXISTS gold.banknotes_features (
            id STRING,
            variance FLOAT,
            skewness FLOAT,
            curtosis FLOAT,
            entropy FLOAT,
            is_fake INT,
            variance_scaled FLOAT,
            skewness_scaled FLOAT,
            curtosis_scaled FLOAT,
            entropy_scaled FLOAT,
            feature_vector ARRAY<FLOAT>,
            ml_ready_timestamp TIMESTAMP
        )
        USING DELTA
        LOCATION '/mnt/gold/banknotes_features'
        COMMENT 'Features preparadas para Machine Learning'
        TBLPROPERTIES (
            'quality' = 'gold',
            'owner' = 'ml_engineering'
        )
    """)

    print("Catálogo de dados inicializado com sucesso!")
```

```python
def create_database_schemas():

    """Cria os bancos de dados necessários"""

    databases = ["bronze", "silver", "gold"]



    for db in databases:

        spark.sql(f"CREATE DATABASE IF NOT EXISTS {db}")

        print(f"✅ Database {db} criado/verificado")



    return True


def setup_storage_paths():

    """Configura os caminhos de storage (simulados para Community Edition)"""

    # Em produção real, usaríamos DBFS ou ADLS


    base_path = "/tmp/banknote_analysis"



    paths = {

        "bronze": f"{base_path}/bronze",

        "silver": f"{base_path}/silver",

        "gold": f"{base_path}/gold"

    }



    import os

    for path in paths.values():

        os.makedirs(path, exist_ok=True)



    print(f"✅ Paths de storage configurados em {base_path}")

    return paths
```

4.3. Pipeline Bronze (Extração de Dados)

```python
class BronzeLayer:
    """Classe para ingestão de dados na camada Bronze"""

    def __init__(self, source_url):
        self.source_url = source_url
        self.batch_id = str(uuid.uuid4())[:8]
        self.ingestion_time = datetime.now()

    def extract_data(self):
        """Extrai dados da fonte"""
        print("📥 Iniciando extração de dados...")

        try:
            # Baixar dados do GitHub
            response = requests.get(self.source_url)
            response.raise_for_status()

            # Converter para DataFrame pandas
            data_lines = response.text.strip().split('\n')
            data_list = []

            for line in data_lines:
                if line.strip():
                    parts = line.strip().split(',')
                    data_list.append({
                        'variance': float(parts[0]),
                        'skewness': float(parts[1]),
                        'curtosis': float(parts[2]),
                        'entropy': float(parts[3]),
                        'class': int(parts[4])
                    })

            # Criar DataFrame Spark
            df = spark.createDataFrame(data_list)

            # Adicionar metadados
            df = df.withColumn("ingestion_timestamp", lit(self.ingestion_time)) \
                .withColumn("source_file", lit(self.source_url)) \
                .withColumn("batch_id", lit(self.batch_id))

            print(f"✅ Extração concluída: {df.count()} registros")
            return df

        except Exception as e:
            print(f"❌ Erro na extração: {str(e)}")
            raise

    def load_to_bronze(self, df):
        """Carrega dados para camada Bronze"""
        print("💾 Carregando para camada Bronze...")

        # Salvar como Delta Table
        bronze_path = "/tmp/banknote_analysis/bronze/banknotes_raw"

        df.write \
            .format("delta") \
            .mode("overwrite") \
            .option("overwriteSchema", "true") \
            .save(bronze_path)

        # Criar tabela no metastore
        spark.sql(f"""
            CREATE TABLE IF NOT EXISTS bronze.banknotes_raw
            USING DELTA
            LOCATION '{bronze_path}'
        """)

        # Atualizar estatísticas - REMOVIDO: ANALYZE TABLE não é suportado para tabelas v2 (Delta Lake)
        # spark.sql(f"ANALYZE TABLE bronze.banknotes_raw COMPUTE STATISTICS")

        print("✅ Dados carregados na camada Bronze")

        # Mostrar preview
        print("📊 Preview da camada Bronze:")
        spark.sql("SELECT * FROM bronze.banknotes_raw LIMIT 5").show()

        return bronze_path
```

4.4. Pipeline Silver (Transformação e Limpeza)

```python
class SilverLayer:

    """Classe para processamento na camada Silver"""



    def __init__(self):

        self.processing_time = datetime.now()



    def load_from_bronze(self):

        """Carrega dados da camada Bronze"""

        print("🔄 Carregando dados da camada Bronze...")



        try:

            df = spark.sql("""

                SELECT

                    variance,

                    skewness,

                    curtosis,

                    entropy,

                    class,

                    ingestion_timestamp,

                    source_file,

                    batch_id

                FROM bronze.banknotes_raw

            """)



            print(f"📥 Dados carregados: {df.count()} registros")

            return df



        except Exception as e:

            print(f"❌ Erro ao carregar da Bronze: {str(e)}")

            raise



    def clean_data(self, df):

        """Aplica limpeza e validação dos dados"""

        print("🧹 Iniciando limpeza de dados...")



        # 1. Remover duplicatas exatas

        initial_count = df.count()

        df = df.dropDuplicates()

        duplicates_removed = initial_count - df.count()



        # 2. Validar domínios

        df = df.filter(

            (col("class").isin([0, 1])) &

            (col("variance").isNotNull()) &

            (col("skewness").isNotNull()) &

            (col("curtosis").isNotNull()) &

            (col("entropy").isNotNull())

        )



        # 3. Renomear colunas para melhor semântica

        df = df.withColumnRenamed("class", "is_fake")



        # 4. Adicionar UUID único

        df = df.withColumn("id", expr("uuid()"))



        # 5. Detectar outliers usando IQR

        features = ["variance", "skewness", "curtosis", "entropy"]



        for feature in features:

            # Calcular quartis

            quantiles = df.approxQuantile(feature, [0.25, 0.75], 0.05)

            q1, q3 = quantiles[0], quantiles[1]

            iqr = q3 - q1



            # Definir limites

            lower_bound = q1 - 1.5 * iqr

            upper_bound = q3 + 1.5 * iqr



            # Marcar outliers

            outlier_col = f"{feature}_outlier"

            df = df.withColumn(

                outlier_col,

                (col(feature) < lower_bound) | (col(feature) > upper_bound)

            )



        # 6. Calcular score de qualidade

        df = df.withColumn(

            "data_quality_score",

            when(

                (col("variance_outlier") == False) &

                (col("skewness_outlier") == False) &

                (col("curtosis_outlier") == False) &

                (col("entropy_outlier") == False),

                100.0

            ).otherwise(80.0)

        )



        # 7. Consolidar flags de outlier

        df = df.withColumn(

            "is_outlier",

            col("variance_outlier") | col("skewness_outlier") |

            col("curtosis_outlier") | col("entropy_outlier")

        )



        # 8. Remover colunas auxiliares

        df = df.drop(*[f"{f}_outlier" for f in features])



        # 9. Adicionar timestamp de processamento

        df = df.withColumn("processing_timestamp", lit(self.processing_time))



        print(f"✅ Limpeza concluída: {duplicates_removed} duplicatas removidas")

        print(f"📊 Registros após limpeza: {df.count()}")



        return df



    def calculate_quality_metrics(self, df):

        """Calcula métricas de qualidade dos dados"""

        print("📈 Calculando métricas de qualidade...")



        metrics = {

            "total_records": df.count(),

            "genuine_notes": df.filter(col("is_fake") == 0).count(),

            "fake_notes": df.filter(col("is_fake") == 1).count(),

            "null_values": {},

            "quality_score_avg": df.agg(avg("data_quality_score")).collect()[0][0],

            "outlier_percentage": df.filter(col("is_outlier") == True).count() / df.count() * 100

        }



        # Contar valores nulos por coluna

        for col_name in ["variance", "skewness", "curtosis", "entropy", "is_fake"]:

            null_count = df.filter(col(col_name).isNull()).count()

            metrics["null_values"][col_name] = null_count



        print("📊 Métricas de Qualidade:")

        for key, value in metrics.items():

            if key != "null_values":

                print(f"   {key}: {value}")



        return metrics



    def load_to_silver(self, df):

        """Carrega dados para camada Silver"""

        print("💾 Carregando para camada Silver...")



        silver_path = "/tmp/banknote_analysis/silver/banknotes_cleaned"



        # Ordenar colunas

        df = df.select(

            "id",

            "variance",

            "skewness",

            "curtosis",

            "entropy",

            "is_fake",

            "data_quality_score",

            "is_outlier",

            "processing_timestamp",

            "ingestion_timestamp"

        )



        # Salvar como Delta Table

        df.write \
            .format("delta") \
            .mode("overwrite") \
            .option("overwriteSchema", "true") \
            .save(silver_path)



        # Criar tabela no metastore

        spark.sql(f"""

            CREATE TABLE IF NOT EXISTS silver.banknotes_cleaned

            USING DELTA

            LOCATION '{silver_path}'

        """)



        # Otimizar tabela

        # spark.sql(f"OPTIMIZE silver.banknotes_cleaned") # Removido para compatibilidade



        print("✅ Dados carregados na camada Silver")



        # Mostrar preview

        print("📊 Preview da camada Silver:")

        spark.sql("""

            SELECT

                id,

                ROUND(variance, 3) as variance,

                ROUND(skewness, 3) as skewness,

                ROUND(curtosis, 3) as curtosis,

                ROUND(entropy, 3) as entropy,

                is_fake,

                ROUND(data_quality_score, 1) as quality_score,

                is_outlier

            FROM silver.banknotes_cleaned

            LIMIT 10

        """).show()



        return silver_path
```

4.5. Pipeline Gold (Enriquecimento para ML e BI)

```python
import math

class GoldLayer:

    """Classe para processamento na camada Gold"""



    def __init__(self):

        self.ml_ready_time = datetime.now()



    def load_from_silver(self):

        """Carrega dados da camada Silver"""

        print("🔄 Carregando dados da camada Silver...")



        df = spark.sql("""

            SELECT

                id,

                variance,

                skewness,

                curtosis,

                entropy,

                is_fake,

                data_quality_score,

                is_outlier

            FROM silver.banknotes_cleaned

            WHERE is_outlier = FALSE  -- Remover outliers para ML

        """)



        print(f"📥 Dados carregados: {df.count()} registros")

        return df



    def calculate_statistical_metrics(self, df):

        """Calcula métricas estatísticas agregadas por classe"""

        print("📊 Calculando métricas estatísticas...")



        features = ["variance", "skewness", "curtosis", "entropy"]

        metrics_list = []



        for feature in features:

            for class_type in [0, 1]:

                class_df = df.filter(col("is_fake") == class_type)



                stats = class_df.select(

                    count(feature).alias("count"),

                    mean(feature).alias("mean"),

                    stddev(feature).alias("std_dev"),

                    min(feature).alias("min"),

                    max(feature).alias("max"),

                    expr(f"percentile_approx({feature}, 0.25)").alias("q1"),

                    expr(f"percentile_approx({feature}, 0.5)").alias("median"),

                    expr(f"percentile_approx({feature}, 0.75)").alias("q3")

                ).collect()[0]



                # Calcular outliers usando IQR

                q1_val = stats["q1"]

                q3_val = stats["q3"]

                iqr = q3_val - q1_val

                lower_bound = q1_val - 1.5 * iqr

                upper_bound = q3_val + 1.5 * iqr



                outlier_count = class_df.filter(

                    (col(feature) < lower_bound) | (col(feature) > upper_bound)

                ).count()



                metrics_list.append({

                    "metric_date": datetime.now().date(),

                    "class_type": class_type,

                    "feature_name": feature,

                    "count": stats["count"],

                    "mean": float(stats["mean"]),

                    "std_dev": float(stats["std_dev"]),

                    "min": float(stats["min"]),

                    "q1": float(stats["q1"]),

                    "median": float(stats["median"]),

                    "q3": float(stats["q3"]),

                    "max": float(stats["max"]),

                    "outlier_count": outlier_count

                })



        # Criar DataFrame de métricas

        metrics_df = spark.createDataFrame(metrics_list)



        print(f"✅ Métricas calculadas: {metrics_df.count()} combinações feature/classe")

        return metrics_df



    def prepare_ml_features(self, df):

        """Prepara features para Machine Learning"""

        print("🤖 Preparando features para ML...")



        from pyspark.ml.feature import StandardScaler, VectorAssembler
        from pyspark.sql.functions import udf
        from pyspark.sql.types import ArrayType, DoubleType
        from pyspark.ml.linalg import VectorUDT

        # Define a UDF to convert VectorUDT to ArrayType
        def vector_to_array(vector):
            if vector is not None:
                return vector.toArray().tolist()
            return None

        vector_to_array_udf = udf(vector_to_array, ArrayType(DoubleType()))

        # Features para escalonar

        features = ["variance", "skewness", "curtosis", "entropy"]



        # 1. Vetorizar features

        assembler = VectorAssembler(

            inputCols=features,

            outputCol="features_vector"

        )

        df_vectorized = assembler.transform(df)



        # 2. Escalonar features (StandardScaler)

        scaler = StandardScaler(

            inputCol="features_vector",

            outputCol="scaled_features",

            withStd=True,

            withMean=True

        )



        scaler_model = scaler.fit(df_vectorized)

        df_scaled = scaler_model.transform(df_vectorized)



        # 3. Extrair features escaladas para colunas individuais
        # Use UDF to convert VectorUDT to ArrayType for element extraction
        for i, feature in enumerate(features):

            df_scaled = df_scaled.withColumn(

                f"{feature}_scaled",

                vector_to_array_udf(col("scaled_features")).getItem(i)

            )



        # 4. Selecionar colunas finais

        df_final = df_scaled.select(

            "id",

            "variance",

            "skewness",

            "curtosis",

            "entropy",

            "is_fake",

            "variance_scaled",

            "skewness_scaled",

            "curtosis_scaled",

            "entropy_scaled",

            col("scaled_features").alias("feature_vector"),

            lit(self.ml_ready_time).alias("ml_ready_timestamp")

        )



        print(f"✅ Features preparadas: {df_final.count()} registros")

        return df_final



    def calculate_correlation_matrix(self, df):

        """Calcula matriz de correlação"""

        print("🔗 Calculando matriz de correlação...")



        features = ["variance", "skewness", "curtosis", "entropy"]



        # Calcular correlações

        correlations = []

        for i in range(len(features)):

            for j in range(i, len(features)):

                if i == j:

                    corr = 1.0

                else:

                    corr = df.stat.corr(features[i], features[j])



                correlations.append({

                    "feature_1": features[i],

                    "feature_2": features[j],

                    "correlation": corr,

                    "abs_correlation": math.fabs(corr) # Use Python's built-in abs() for float

                })



        # Criar DataFrame de correlações

        corr_df = spark.createDataFrame(correlations)



        print("📊 Matriz de correlação calculada")

        corr_df.show()



        return corr_df



    def load_to_gold(self, metrics_df, features_df):

        """Carrega dados para camada Gold"""

        print("💾 Carregando para camada Gold...")



        # 1. Carregar métricas

        metrics_path = "/tmp/banknote_analysis/gold/banknotes_metrics"



        metrics_df.write \
            .format("delta") \
            .mode("overwrite") \
            .option("overwriteSchema", "true") \
            .partitionBy("metric_date", "class_type") \
            .save(metrics_path)



        spark.sql(f"""

            CREATE TABLE IF NOT EXISTS gold.banknotes_metrics

            USING DELTA

            LOCATION '{metrics_path}'

        """)



        # 2. Carregar features para ML

        features_path = "/tmp/banknote_analysis/gold/banknotes_features"



        features_df.write \
            .format("delta") \
            .mode("overwrite") \
            .option("overwriteSchema", "true") \
            .save(features_path)



        spark.sql(f"""

            CREATE TABLE IF NOT EXISTS gold.banknotes_features

            USING DELTA

            LOCATION '{features_path}'

        """)



        # 3. Otimizar tabelas

        # spark.sql("OPTIMIZE gold.banknotes_metrics") # Removido para compatibilidade

        # spark.sql("OPTIMIZE gold.banknotes_features") # Removido para compatibilidade



        print("✅ Dados carregados na camada Gold")



        # Mostrar previews

        print("📊 Preview das métricas:")

        spark.sql("""

            SELECT

                feature_name,

                class_type,

                ROUND(mean, 3) as mean,

                ROUND(std_dev, 3) as std,

                ROUND(median, 3) as median,

                outlier_count

            FROM gold.banknotes_metrics

            WHERE metric_date = CURRENT_DATE()

            ORDER BY feature_name, class_type

        """).show()



        print("📊 Preview das features para ML:")

        spark.sql("""

            SELECT

                id,

                ROUND(variance_scaled, 3) as var_scaled,

                ROUND(skewness_scaled, 3) as skew_scaled,

                ROUND(curtosis_scaled, 3) as curt_scaled,

                ROUND(entropy_scaled, 3) as ent_scaled,

                is_fake

            FROM gold.banknotes_features

            LIMIT 10

        """).show()



        return metrics_path, features_path
```

4.6. Pipeline Principal

```python
class BanknoteDataPipeline:

    """Pipeline principal que orquestra todas as camadas"""



    def __init__(self, source_url):

        self.source_url = source_url

        self.results = {}



    def run(self):

        """Executa a pipeline completa"""

        print("=" * 60)

        print("🚀 INICIANDO PIPELINE DE DADOS - DETECÇÃO DE CÉDULAS FALSAS")

        print("=" * 60)



        # 1. Configuração inicial

        print("\n1️⃣ CONFIGURAÇÃO INICIAL")

        create_database_schemas()

        paths = setup_storage_paths()



        # 2. Camada Bronze

        print("\n2️⃣ CAMADA BRONZE - EXTRAÇÃO")

        bronze_layer = BronzeLayer(self.source_url)

        bronze_df = bronze_layer.extract_data()

        bronze_path = bronze_layer.load_to_bronze(bronze_df)

        self.results["bronze"] = bronze_path



        # 3. Camada Silver

        print("\n3️⃣ CAMADA SILVER - TRANSFORMAÇÃO")

        silver_layer = SilverLayer()

        silver_source_df = silver_layer.load_from_bronze()

        cleaned_df = silver_layer.clean_data(silver_source_df)

        quality_metrics = silver_layer.calculate_quality_metrics(cleaned_df)

        silver_path = silver_layer.load_to_silver(cleaned_df)

        spark.catalog.refreshTable("silver.banknotes_cleaned") # Adicionado: Refrescar catálogo para reconhecer a nova tabela

        self.results["silver"] = {

            "path": silver_path,

            "quality_metrics": quality_metrics

        }



        # 4. Camada Gold

        print("\n4️⃣ CAMADA GOLD - ENRIQUECIMENTO")

        gold_layer = GoldLayer()

        gold_source_df = gold_layer.load_from_silver()



        # Calcular métricas estatísticas

        metrics_df = gold_layer.calculate_statistical_metrics(gold_source_df)



        # Preparar features para ML

        features_df = gold_layer.prepare_ml_features(gold_source_df)



        # Calcular correlações

        corr_df = gold_layer.calculate_correlation_matrix(gold_source_df)



        # Carregar para Gold

        metrics_path, features_path = gold_layer.load_to_gold(metrics_df, features_df)



        self.results["gold"] = {

            "metrics_path": metrics_path,

            "features_path": features_path,

            "correlations": corr_df

        }



        # 5. Relatório Final

        print("\n" + "=" * 60)

        print("📋 RELATÓRIO FINAL DA PIPELINE")

        print("=" * 60)



        self.generate_report()



        return self.results



    def generate_report(self):

        """Gera relatório final da execução"""

        print(f"\n📊 RESUMO DA EXECUÇÃO:")

        print(f"   • Camada Bronze: {spark.sql('SELECT COUNT(*) FROM bronze.banknotes_raw').collect()[0][0]} registros")

        print(f"   • Camada Silver: {spark.sql('SELECT COUNT(*) FROM silver.banknotes_cleaned').collect()[0][0]} registros")

        print(f"   • Camada Gold - Métricas: {spark.sql('SELECT COUNT(*) FROM gold.banknotes_metrics').collect()[0][0]} linhas")

        print(f"   • Camada Gold - Features: {spark.sql('SELECT COUNT(*) FROM gold.banknotes_features').collect()[0][0]} registros")



        # Distribuição das classes

        dist = spark.sql("""

            SELECT

                CASE WHEN is_fake = 0 THEN 'Genuína' ELSE 'Falsa' END as tipo,

                COUNT(*) as quantidade,

                ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) as percentual

            FROM silver.banknotes_cleaned

            GROUP BY is_fake

            ORDER BY is_fake

        """).collect()



        print(f"\n🎯 DISTRIBUIÇÃO DAS CLASSES:")

        for row in dist:

            print(f"   • {row['tipo']}: {row['quantidade']} ({row['percentual']}%) ")



        # Métricas de qualidade

        quality = self.results["silver"]["quality_metrics"]

        print(f"\n🏆 QUALIDADE DOS DADOS:")

        print(f"   • Score médio: {quality['quality_score_avg']:.2f}")

        print(f"   • Outliers: {quality['outlier_percentage']:.2f}%")



        print(f"\n✅ PIPELINE EXECUTADA COM SUCESSO!")

        print("=" * 60)
```

4.8. Análise Exploratória Adicional

```python
def perform_eda():
    """Realiza Análise Exploratória de Dados"""
    print("\n🔍 ANÁLISE EXPLORATÓRIA DE DADOS")
    print("=" * 40)

    # 1. Correlação entre features
    print("1. MATRIZ DE CORRELAÇÃO:")
    corr_matrix = spark.sql("""
        SELECT
            feature_1,
            feature_2,
            ROUND(correlation, 3) as correlation,
            CASE
                WHEN ABS(correlation) > 0.7 THEN 'Forte'
                WHEN ABS(correlation) > 0.3 THEN 'Moderada'
                ELSE 'Fraca'
            END as magnitude
        FROM (
            SELECT
                f1.feature_name as feature_1,
                f2.feature_name as feature_2,
                CORR(f1.value, f2.value) as correlation
            FROM (
                SELECT
                    id,
                    'variance' as feature_name,
                    variance as value
                FROM silver.banknotes_cleaned
                UNION ALL
                SELECT
                    id,
                    'skewness' as feature_name,
                    skewness as value
                FROM silver.banknotes_cleaned
                UNION ALL
                SELECT
                    id,
                    'curtosis' as feature_name,
                    curtosis as value
                FROM silver.banknotes_cleaned
                UNION ALL
                SELECT
                    id,
                    'entropy' as feature_name,
                    entropy as value
                FROM silver.banknotes_cleaned
            ) f1
            JOIN (
                SELECT
                    id,
                    'variance' as feature_name,
                    variance as value
                FROM silver.banknotes_cleaned
                UNION ALL
                SELECT
                    id,
                    'skewness' as feature_name,
                    skewness as value
                FROM silver.banknotes_cleaned
                UNION ALL
                SELECT
                    id,
                    'curtosis' as feature_name,
                    curtosis as value
                FROM silver.banknotes_cleaned
                UNION ALL
                SELECT
                    id,
                    'entropy' as feature_name,
                    entropy as value
                FROM silver.banknotes_cleaned
            ) f2 ON f1.id = f2.id AND f1.feature_name <= f2.feature_name
            GROUP BY f1.feature_name, f2.feature_name
        )
        WHERE feature_1 != feature_2
        ORDER BY ABS(correlation) DESC
    """)

    corr_matrix.show()

    # 2. Features mais discriminativas
    print("\n2. FEATURES MAIS DISCRIMINATIVAS:")
    discriminative_features = spark.sql("""
        WITH stats AS (
            SELECT
                feature_name,
                class_type,
                mean,
                std_dev
            FROM gold.banknotes_metrics
            WHERE metric_date = CURRENT_DATE()
        ),
        diff AS (
            SELECT
                s1.feature_name,
                ABS(s1.mean - s0.mean) / ((s1.std_dev + s0.std_dev) / 2) as separability_score
            FROM stats s1
            JOIN stats s0 ON s1.feature_name = s0.feature_name
            WHERE s1.class_type = 1 AND s0.class_type = 0
        )
        SELECT
            feature_name,
            ROUND(separability_score, 3) as separability,
            CASE
                WHEN separability_score > 1.0 THEN 'Alta'
                WHEN separability_score > 0.5 THEN 'Média'
                ELSE 'Baixa'
            END as discriminative_power
        FROM diff
        ORDER BY separability_score DESC
    """)

    discriminative_features.show()

    # 3. Distribuição por quartis
    print("\n3. DISTRIBUIÇÃO POR QUARTIS (Genuínas vs Falsas):")
    for feature in ["variance", "curtosis"]:
        print(f"\n   📊 {feature.upper()}:")
        stats = spark.sql(f"""
            SELECT
                CASE WHEN is_fake = 0 THEN 'Genuína' ELSE 'Falsa' END as tipo,
                ROUND(MIN({feature}), 3) as min,
                ROUND(PERCENTILE({feature}, 0.25), 3) as q1,
                ROUND(PERCENTILE({feature}, 0.5), 3) as mediana,
                ROUND(PERCENTILE({feature}, 0.75), 3) as q3,
                ROUND(MAX({feature}), 3) as max
            FROM silver.banknotes_cleaned
            WHERE is_outlier = FALSE
            GROUP BY is_fake
            ORDER BY is_fake
        """)
        stats.show()

# Executar análise exploratória
perform_eda()
```

```python
def monitor_pipeline_health():
    """Monitora a saúde da pipeline"""
    print("\n🏥 MONITORAMENTO DA PIPELINE")
    print("=" * 40)

    checks = []

    # 1. Verificar completude das tabelas
    tables = [
        ("bronze.banknotes_raw", "Tabela Bronze"),
        ("silver.banknotes_cleaned", "Tabela Silver"),
        ("gold.banknotes_metrics", "Tabela Gold - Métricas"),
        ("gold.banknotes_features", "Tabela Gold - Features")
    ]

    for table, name in tables:
        try:
            count = spark.sql(f"SELECT COUNT(*) FROM {table}").collect()[0][0]
            checks.append((name, "✅", f"{count} registros"))
        except:
            checks.append((name, "❌", "Tabela não encontrada"))

    # 2. Verificar qualidade dos dados
    quality_check = spark.sql("""
        SELECT
            ROUND(AVG(data_quality_score), 2) as avg_quality,
            ROUND(SUM(CASE WHEN is_outlier THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) as outlier_pct
        FROM silver.banknotes_cleaned
    """).collect()[0]

    checks.append(("Qualidade Média", "✅" if quality_check["avg_quality"] > 90 else "⚠️", f"{quality_check['avg_quality']}%"))
    checks.append(("Outliers", "✅" if quality_check["outlier_pct"] < 5 else "⚠️", f"{quality_check['outlier_pct']}%"))

    # 3. Verificar balanceamento
    balance_check = spark.sql("""
        SELECT
            ROUND(MIN(percentual), 2) as min_pct,
            ROUND(MAX(percentual), 2) as max_pct
        FROM (
            SELECT
                is_fake,
                COUNT(*) * 100.0 / SUM(COUNT(*)) OVER() as percentual
            FROM silver.banknotes_cleaned
            GROUP BY is_fake
        )
    """).collect()[0]

    balance_ratio = balance_check["min_pct"] / balance_check["max_pct"]
    checks.append(("Balanceamento", "✅" if balance_ratio > 0.7 else "⚠️", f"Ratio: {balance_ratio:.2f}"))

    # Exibir resultados
    for check_name, status, details in checks:
        print(f"{status} {check_name}: {details}")

# Executar monitoramento
monitor_pipeline_health()
```

```python
def generate_simple_dashboard():
    """Gera um dashboard simples com métricas principais"""
    print("\n📈 DASHBOARD DE MÉTRICAS")
    print("=" * 50)

    import matplotlib.pyplot as plt
    import numpy as np
    import math # Import math explicitly for fabs

    # Coletar dados para o dashboard
    dashboard_data = {}

    # 1. Distribuição de classes
    class_dist = spark.sql("""
        SELECT
            CASE WHEN is_fake = 0 THEN 'Genuína' ELSE 'Falsa' END as classe,
            COUNT(*) as quantidade
        FROM silver.banknotes_cleaned
        GROUP BY is_fake
    """).toPandas()

    # 2. Métricas por feature
    feature_metrics = spark.sql("""
        SELECT
            feature_name,
            class_type,
            mean,
            std_dev
        FROM gold.banknotes_metrics
        WHERE metric_date = CURRENT_DATE()
        ORDER BY feature_name, class_type
    """).toPandas()

    # 3. Correlações
    correlations = spark.sql("""
        SELECT
            feature_1,
            feature_2,
            correlation
        FROM (
            SELECT
                f1.feature_name as feature_1,
                f2.feature_name as feature_2,
                CORR(f1.value, f2.value) as correlation
            FROM (
                SELECT 'variance' as feature_name, variance as value FROM silver.banknotes_cleaned
                UNION ALL SELECT 'skewness', skewness FROM silver.banknotes_cleaned
                UNION ALL SELECT 'curtosis', curtosis FROM silver.banknotes_cleaned
                UNION ALL SELECT 'entropy', entropy FROM silver.banknotes_cleaned
            ) f1
            JOIN (
                SELECT 'variance' as feature_name, variance as value FROM silver.banknotes_cleaned
                UNION ALL SELECT 'skewness', skewness FROM silver.banknotes_cleaned
                UNION ALL SELECT 'curtosis', curtosis FROM silver.banknotes_cleaned
                UNION ALL SELECT 'entropy', entropy FROM silver.banknotes_cleaned
            ) f2 ON f1.feature_name <= f2.feature_name
            GROUP BY f1.feature_name, f2.feature_name
        )
        WHERE feature_1 != feature_2
    """).toPandas()

    # Exibir dashboard textual
    print("\n🎯 RESUMO EXECUTIVO:")
    print(f"   • Total de cédulas analisadas: {class_dist['quantidade'].sum()}")
    print(f"   • Genuínas: {class_dist.loc[class_dist['classe'] == 'Genuína', 'quantidade'].values[0]}")
    print(f"   • Falsas: {class_dist.loc[class_dist['classe'] == 'Falsa', 'quantidade'].values[0]}")

    print("\n📊 FEATURES MAIS DISCRIMINATIVAS:")
    for feature in ['variance', 'curtosis', 'entropy', 'skewness']:
        genuine_mean = feature_metrics[(feature_metrics['feature_name'] == feature) &
                                      (feature_metrics['class_type'] == 0)]['mean'].values[0]
        fake_mean = feature_metrics[(feature_metrics['feature_name'] == feature) &
                                   (feature_metrics['class_type'] == 1)]['mean'].values[0]
        diff = math.fabs(genuine_mean - fake_mean) # Use math.fabs for Python floats
        print(f"   • {feature.capitalize()}: Diferença média = {diff:.3f}")

    print("\n🔗 CORRELAÇÕES PRINCIPAIS:")
    top_correlations = correlations.nlargest(3, 'correlation')
    for _, row in top_correlations.iterrows():
        print(f"   • {row['feature_1']} × {row['feature_2']}: {row['correlation']:.3f}")

    print("\n⚠️ ALERTAS:")
    # Verificar outliers
    outlier_pct = spark.sql("""
        SELECT ROUND(SUM(CASE WHEN is_outlier THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) as pct
        FROM silver.banknotes_cleaned
    """).collect()[0][0]

    if outlier_pct > 5:
        print(f"   • Alto percentual de outliers: {outlier_pct}%")

    # Verificar qualidade
    avg_quality = spark.sql("SELECT AVG(data_quality_score) FROM silver.banknotes_cleaned").collect()[0][0]
    if avg_quality < 90:
        print(f"   • Qualidade dos dados abaixo do ideal: {avg_quality:.1f}%")

# Gerar dashboard
generate_simple_dashboard()
```
